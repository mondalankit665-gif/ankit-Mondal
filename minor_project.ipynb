```python
import pandas as pd
import matplotlib.pyplot as plt
import streamlit as st
import numpy as np

# 1. Define the data
data = {
    'Date': pd.date_range(start='2023-01-01', periods=10, freq='D'),
    'Amount': [120, 150, 80, 200, 170, 90, 110, 130, 160, 140]
}
df = pd.DataFrame(data)
income = 5000

# 2. Visualization
fig, ax = plt.subplots(figsize=(10, 5))
df.groupby('Date')['Amount'].sum().plot(kind='line', ax=ax, marker='o')
ax.set_title('Expense Trend Over Time')
ax.set_ylabel('Amount ($)')
ax.grid(True)

# Handle Streamlit display (if running in a streamlit context)
try:
    st.title("Expense Tracker")
    st.pyplot(fig)
except Exception:
    # Fallback for standard notebook execution
    plt.show()

# 3. Calculation and Output
avg_expenses = df['Amount'].sum() / len(df)
predicted_savings = income - (avg_expenses * 30) # Simple monthly estimate

print(f"Average Daily Expense: ${avg_expenses:.2f}")
print(f"Predicted Monthly Savings: ${predicted_savings:.2f}")

try:
    st.write(f"Predicted Monthly Savings: ${predicted_savings:.2f}")
except:
    pass
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import streamlit as st
import numpy as np

# Create sample data so the code can run
data = {
    'Date': pd.date_range(start='2023-01-01', periods=10, freq='D'),
    'Amount': [120, 150, 80, 200, 170, 90, 110, 130, 160, 140]
}
df = pd.DataFrame(data)
income = 5000

# Trend plot
fig2, ax2 = plt.subplots()
df.groupby("Date")["Amount"].sum().plot(kind="line", ax=ax2)
ax2.set_title("Expense Trend")
st.pyplot(fig2)
plt.show()

# Savings prediction (simple)
avg_expenses = df.groupby("Date")["Amount"].sum().mean()
predicted_savings = income - avg_expenses
st.write(f"Predicted Monthly Savings: ${predicted_savings:.2f}")
print(f"Predicted Monthly Savings: ${predicted_savings:.2f}")
```
